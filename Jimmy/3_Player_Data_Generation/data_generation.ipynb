In this notebook, we use the elo rating system to generate statistics of players from past matches.  
Columns: ['player1', 'player2', 'best_of', 'player1_elo', 'player2_elo'
                               'elo_match_win_rate', 'elo_frame_win_rate',
                                'p1_matches_played', 'p1_matches_won', 'p1_frames_played', 'p1_frames_won', 
                                'p2_matches_played', 'p2_matches_won', 'p2_frames_played', 'p2_frames_won', 
                                'p1_frames_played_1_year', 'p1_frames_won_1_year', 'p1_frames_played_3_years', 'p1_frames_won_3_years',
                                'p2_frames_played_1_year','p2_frames_won_1_year', 'p2_frames_played_3_years','p2_frames_won_3_years',
                                'score1', 'score2','match_result', 'win_percentage']

```python
import pandas as pd
import numpy as np
```

```python
# Import professional snooker matches data form 1982 to 2025
matches = pd.read_csv('/Users/tliu/Desktop/Erdos Project/1_Match_Data/matches.csv')
matches.head()
```

```python
from elo_v4 import elo_rating

#Store all the tournaments in a list
tournaments = list(matches.tournament_id.unique())
#nth means nth tournament among 1085 tournaments
nth = range(985,1035)


data = pd.DataFrame(columns = ['player1', 'player2', 'best_of', 'player1_elo', 'player2_elo',
                               'elo_match_win_rate', 'elo_frame_win_rate',
                                'p1_matches_played', 'p1_matches_won', 'p1_frames_played', 'p1_frames_won', 
                                'p2_matches_played', 'p2_matches_won', 'p2_frames_played', 'p2_frames_won', 
                                'p1_frames_played_1_year', 'p1_frames_won_1_year', 'p1_frames_played_3_years', 'p1_frames_won_3_years',
                                'p2_frames_played_1_year','p2_frames_won_1_year', 'p2_frames_played_3_years','p2_frames_won_3_years',
                                'score1', 'score2','match_result', 'win_percentage', 'tournament_id'])



for n in nth:
  past_tournaments = tournaments[:n]
  print(f'Generating data for the {n}th tournament')
  current = tournaments[n]
  past_index = matches['tournament_id'].map(lambda x: x in past_tournaments)
  
  matches_past = matches[past_index]
  matches_future = matches[matches['tournament_id'] == current]
  matches_future_X = matches_future[['player1', 'player2', 'best_of']]

  #Update statistics of players
  elo = elo_rating()
  elo.update_stats(matches = matches_past)
  stats = elo.stats
  sby = elo.stats_by_years
  player_stats = elo.generate_stats(matches_future_X)

  results = np.zeros(len(matches_future))
  win_perc = matches_future['score1'] / (matches_future['score1'] + matches_future['score2'])

  
  new_data = player_stats
  new_data['score1'] = matches_future['score1'].values
  new_data['score2'] = matches_future['score2'].values
  new_data['match_result'] = results
  new_data['win_percentage'] = win_perc.values
  new_data['tournament_id'] = tournaments[n]
  

  data = pd.concat([data, new_data], axis = 0, ignore_index = True)
  






```
