In the match data generated in data_generation.ipynb, player1 is always the winner. For the purpose of future data exploration and modeling, we exchange player1 and player2 for half of the matches and save it in another csv file.

```python
import pandas as pd
import numpy as np
import random
```

```python
data = pd.read_csv('match_data_20_tourns.csv')
```

```python
data.head()
```

```python
row = data.loc[0]
```

```python
#We use random.shuffle to choose the indices of matches we want to modify.
match_index = list(range(len(data)))
random.shuffle(match_index)
modify_index = match_index[:len(data)//2]
```

```python
modify_index[:5]
```

```python
data.loc[modify_index].head(5)
```

```python
#Create a new dataframe df to store the modified data
df = data.loc[modify_index]
```

```python
data.columns
```

```python
#Exchange the stats of player1 and player2
df['player1'], df['player2'] = df['player2'], df['player1']
df['player1_elo'], df['player2_elo'] = df['player2_elo'], df['player1_elo']
df['elo_match_win_rate'], df['elo_frame_win_rate'] = 1-df['elo_match_win_rate'], 1-df['elo_frame_win_rate']
df['p1_matches_played'], df['p1_matches_won'], df['p2_matches_played'], df['p2_matches_won'] = df['p2_matches_played'], df['p2_matches_won'], df['p1_matches_played'], df['p1_matches_won']
df['p1_frames_played'], df['p1_frames_won'], df['p2_frames_played'], df['p2_frames_won'] = df['p2_frames_played'], df['p2_frames_won'], df['p1_frames_played'], df['p1_frames_won']
df['p1_frames_played_1_year'], df['p1_frames_won_1_year'], df['p2_frames_played_1_year'], df['p2_frames_won_1_year'] = df['p2_frames_played_1_year'], df['p2_frames_won_1_year'], df['p1_frames_played_1_year'], df['p1_frames_won_1_year']
df['p1_frames_played_3_years'], df['p1_frames_won_3_years'], df['p2_frames_played_3_years'], df['p2_frames_won_3_years'] = df['p2_frames_played_3_years'], df['p2_frames_won_3_years'], df['p1_frames_played_3_years'], df['p1_frames_won_3_years']
df['score1'], df['score2'] = df['score2'], df['score1']
df['match_result'] = 1
df['win_percentage'] = 1 - df['win_percentage']
```

```python
data.loc[modify_index] = df
```
