```python
def date_format(s):
       
    if isinstance(s, str):
        return pd.to_datetime(s[:10], format = '%Y-%m-%d')
        print(s) 
    else:
        return None
        
```

```python
matches = pd.read_csv('/Users/tliu/Desktop/Erdos Project/1_Match_Data/matches.csv')
matches = matches[-len(data):]
dates = matches['date'].apply(date_format)
data['date'] = dates.values
```
