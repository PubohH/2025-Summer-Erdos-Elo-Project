```python
import pandas as pd
import numpy as np
```

```python
# Import professional snooker matches data form 1982 to 2025
matches = pd.read_csv('/Users/tliu/Desktop/Erdos Project/Data/matches.csv')
matches.head()
```

```python
#Predict the match result, expected frame win rate and matches win rate of the 1080th tournament
from elo_v4 import elo_rating
tournaments = list(matches.tournament_id.unique())
past_tournaments = tournaments[:1080]
current = tournaments[1080]
past_index = matches['tournament_id'].map(lambda x: x in past_tournaments)

matches_past = matches[past_index]
matches_future = matches[matches['tournament_id'] == current]

elo = elo_rating()
elo.update_stats(matches = matches_past)

prediction , frames_win_rates, match_win_rates = elo.predict(matches_future[['player1', 'player2', 'best_of']])


matches_future.loc[:, 'prediction'] = prediction
matches_future.loc[:, 'frames_win_rates'] = frames_win_rates
matches_future.loc[:, 'match_win_rates'] = match_win_rates

```
