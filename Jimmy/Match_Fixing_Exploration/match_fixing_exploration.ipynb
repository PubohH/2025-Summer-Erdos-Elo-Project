Attempt to find players who involved in match fixing by looking at those who lose frequent to players with much lower elo-rating.

```python
import pandas as pd
import numpy as np
```

```python
matches = pd.read_csv('/Users/tliu/Desktop/Erdos Project/Data/matches.csv')
```

```python
from elo import elo_rating
elo = elo_rating()
elo.update_stats(matches = matches, difference = 50)
stats = elo.stats
```

```python
stats
```

```python
stats = stats.sort_values(by = 'big_loss', ascending=False)
stats
```

```python
stats['big_loss_perc'] = stats['big_loss']/stats['matches_played']
stats1 = stats.sort_values(by = 'big_loss_perc', ascending=False)
stats1[:20]
```

```python
stats2 = stats[stats['matches_played']>200]
stats2[:20]
```
